```python
import pandas as pd
```

**Method 1: From a Dictionary**

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'London', 'Tokyo'],
    'Salary': [75000, 85000, 90000, 79000]
}

df = pd.DataFrame(data)
print(df)
```

**Method 2: From a list of Lists**

```python
df2 = pd.DataFrame(
    [['Alice', 25], ['Bob', 30]],
    columns = ['Name', 'age']
)

print(df2)
```
