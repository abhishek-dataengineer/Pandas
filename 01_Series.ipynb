```python
import pandas as pd
```

 **Creating a Series from a list**

```python
marks = pd.Series([85, 90, 78, 92, 88])
print(marks)
```

**Creating a Series with custom Index**

```python
marks = pd.Series([85, 90, 78, 92, 88],
                  index=['Alice', 'Bob', 'Charlie', 'David', 'Eve'])
print(marks)

print(marks['Alice'])
```

**Creating from a Dictionary**

```python
prices = pd.Series({"Apple": 150.0, "Kiwi": 200.0, "Tesla": 900.5})
print(prices)
```

**Series Operations**

```python
print(prices.mean())
```

```python
print(prices.max())
```

```python
print(marks.mean())
print(marks.max())
```

```python
print(marks.min())
print(prices.min())
```

```python
print(marks.sum())
print(prices.sum())
```

```python
print(marks>80)
```
